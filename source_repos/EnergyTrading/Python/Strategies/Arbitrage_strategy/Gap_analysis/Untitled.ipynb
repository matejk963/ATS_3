```python
def trd_gap_calc(row):
    if bool((row['trd_price'] > (row['a_price'])) or (row['trd_price'] < (row['b_price']))):
        if row['trd_side'] > 0.5:
            return row['trd_price'] - row['a_price']
        else:
            return row['trd_price'] - row['b_price']
    else:
        return 0.0

df_data['trd_gap'] = df_data.apply(trd_gap_calc, axis=1)

trd_gaps = df_data[(abs(df_data['trd_gap']) > 0.051) & (((df_data['trd_gap']>0) &(df_data['ask_t1']))|((df_data['trd_gap']<0) & (df_data['bid_t1'])))]['trd_gap'].apply(abs).apply(lambda x: round(x,2))
#trd_gaps = df_data[(abs(df_data['trd_gap']) > 0.051)]['trd_gap'].apply(abs).apply(lambda x: round(x,2))
```

```python
a=pd.DataFrame(trd_gaps.value_counts()).reset_index()

b=a[a['trd_gap']<0.26]
b['opportunity']=b['count']*(b['trd_gap']-0.05)*700*3

sum(b['count']*(b['trd_gap']-0.05)*700)*3
```
