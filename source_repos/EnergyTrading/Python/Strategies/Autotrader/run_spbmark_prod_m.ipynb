# Steering

```python
from abc import ABC
import inspect

class SteeringTemplate(ABC):
    def get_steering_functions(self):
        steering_functions = {}
        for name, method in inspect.getmembers(self, predicate=inspect.ismethod):
            if name.startswith("steering_"):
                steering_functions[name] = method
        return steering_functions
```

```python
class SpBMarkSteering(SteeringTemplate):

    def steering_stats(self):
        return 'stats'
    
    def steering_close_make_best(self):
        return {'close_behavior': 'makebest'}
    
    def steering_close_aggress(self):
        return {'close_behavior': 'aggress'}
    
    def steering_close_wait(self):
        return {'close_behavior': 'wait'}
    
    def steering_close_auto(self):
        return {'close_behavior': 'auto'}
    
    def steering_open_price(self, value):
        if isinstance(value, float):
            return {'open_price': value}
        else:
            raise ValueError("Argument 'value' must be of type float")
    
    def steering_make_profit_margin(self, value):
        if isinstance(value, float):
            return {'make_profit_margin': value}
        else:
            raise ValueError("Argument 'value' must be of type float")
```
