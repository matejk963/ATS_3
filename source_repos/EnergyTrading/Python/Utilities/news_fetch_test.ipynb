```python
import refinitiv.data as rd
from bs4 import BeautifulSoup
```

```python
rd.open_session()
```

```python
test = rd.news.get_headlines('CFI2Z5',
                      start='2025-03-17',
                      end='2025-03-18',
                      count=1000)
```

```python
rd.open_session()
news_dict = {}
ids_unsuccesful = []
for i,row in test.iterrows():
    story_id = row['storyId']
    try:
        response = rd.content.news.story.Definition(story_id).get_data()
        soup = BeautifulSoup(response.data.story.content.html, "html.parser")
        # Extract text
        readable_text = soup.get_text(separator="\n")
        news_dict[story_id] = readable_text
    except:
        ids_unsuccesful.append(story_id)
        continue
```
